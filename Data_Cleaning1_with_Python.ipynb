```python
import pandas as pd
import numpy as np
```

```python


np.random.seed(42) #Supaya Hasilnya Sama Semua

n = 100
data = pd.DataFrame({
    'id_transaksi': range(1, n+1),
    'nama': np.random.choice(['Budi', 'Ani', 'Citra', 'Doni', 'Eka'], n),
    'umur': np.random.randint(18, 60, n).astype(float),
    'kota': np.random.choice(['sby', ' SBY', 'Surabaya ', 'jkt', 'Jakarta'], n),
    'jenis_pembayaran': np.random.choice(['Transfer', 'CASH', 'QRIS', 'transfer'], n),
    'total_belanja': np.random.randint(50000, 2000000, n).astype(float),
    'tanggal': pd.date_range('2025-09-01', periods=n).astype(str)
})

# 1. Kotorkan Data: Kasih NaN
data.loc[np.random.choice(n, 10, replace=False), 'umur'] = np.nan
data.loc[np.random.choice(n, 8, replace=False), 'total_belanja'] = np.nan

# 2. Kotorkan Data: Kasih Duplikat 5 baris
data = pd.concat([data, data.sample(5)], ignore_index=True)

# 3. Kotorkan Data: Tipe data & format aneh
data['total_belanja'] = data['total_belanja'].apply(lambda x: f"Rp{x:,.0f}".replace(',', '.') if pd.notna(x) else x)
data.loc[3, 'umur'] = 199 # Outlier
data.loc[7, 'total_belanja'] = 'Rp-50.000' # Outlier negatif

# 4. Simpan Jadi CSV
data.to_csv('data_penjualan_kotor.csv', index=False)
print("Dataset 100+ baris wes dadi: data_penjualan_kotor.csv")
print(data.head())
```

```python
#Ubah Tipe data umur, total belanja, dan tanggal

#umur
data['umur']=data['umur'].astype('Int64')

#hilangkan Rp di total belanja
data['total_belanja'] = data['total_belanja'].astype(str)
data['total_belanja'] = data['total_belanja'].replace('nan', np.nan) # Convert string 'nan' back to np.nan

# Now, perform string replacements. Use regex=False for literal matching of '.' and ','.
data['total_belanja'] = data['total_belanja'].str.replace('Rp', '', regex=False)
data['total_belanja'] = data['total_belanja'].str.replace('.', '', regex=False)
data['total_belanja'] = data['total_belanja'].str.replace(',', '', regex=False)

#tanggal
# Convert 'tanggal' column to datetime objects
data['tanggal'] = pd.to_datetime(data['tanggal'])
```

```python
data['total_belanja'] = pd.to_numeric(data['total_belanja'], errors='coerce')
data['total_belanja'] = data['total_belanja'].astype('Int64')
```
